# Intersection
### The Intersection operation returns the common elements present in two or more sets.




```python
A = {1,2,3,4}
B = {3,4,5,6}

print(A & B)
print(A.intersection(B))
```

```python
A = {"Python", "Java", "C++"}
B = {"Java", "C"}

print(A.intersection(B))
```

```python
A = {1,2,3,4}
B = {2,3,5}
C = {2,3,6}

print(A.intersection(B,C))
```

```python
set_1 = {"a", "b", "c"}
set_2 = {"c", "d", "a"}
set_3 = {"d", "g", "c"}

# Finding intersection using intersection() method:
print("Intersection of three sets is:", set_1.intersection(set_2, set_3))
```

## Set Relation Methods
#### issubset() and issuperset() are called set relationship methods because they compare relationships between sets.”

## issubset()
##### checks whether all elements of one set are present in another set.


```python
#Check whether set A is a subset of B
```

```python
A={1,2,3,4,5}
B={4,5,6,7,8}
print (A.issubset(B)) #Output: False
print (A<=B)#Output: False

```

```python
A={1,2,3}
B={1,2,3,4,5}
print (A.issubset(B)) #Output: True
print (A<=B)#Output: #true
```

# isdisjoint()
#### Returns True if the set has no elements in common with the other. Sets are disjointed if and only if their intersection is the empty set.


```python
#Set A and Set B containing common elements
A={1,2,3,4,5}
B={4,5,6,7,8}
print (A.isdisjoint(B))

#Set A and Set B not containing common elements
A={1,2}
B={3,4}
print (A.isdisjoint(B))
```
